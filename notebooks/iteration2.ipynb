## 1. Model Scope

- Full end-to-end autism pathway is modeled.
- Simulation time unit is **days**.
- Resources are currently modeled as **24/7 available capacity** (no explicit weekday/weekend calendar constraint).

## 2. Model Assumptions

- Arrivals follow an exponential process (`IAT = 1 / REFERRALS_PER_DAY`).
- Stage service times follow triangular distributions using the configured min/mode/max values.
- Each stage has its own dedicated resource pool (no cross-cover between stages).
- Resource capacity is constant over time (no sickness, leave, shift variation, or overtime).
- No explicit weekend/holiday shutdown; services run continuously in model time.
- No explicit DNA/cancellation/rebooking loop is modeled.
- Patients move forward through the pathway or exit at defined gatekeeping points.
- Branching probabilities are fixed during a run (no time-varying policy changes).
- Patients are served first-come, first-served within each stage resource queue.
- Run-level randomness is controlled by fixed random streams for reproducibility.
- Audit class is not fully developed and waiting for correct resource modeling

## 3. Testing Coverage

- Includes deterministic boundary tests for pathway exits and progression.
- Covers both **100% exit** and **0% exit** boundaries at key stages (triage, screening, pre-assessment, assessment, and further assessment).

## 4. Analytical Validation

- Includes mathematically derived expected patient counts per stage using branching probabilities.
- Compares analytical expectations against replicated simulation outputs to validate stage-level flow behavior.

## 6. Experiment Setup

Parameter container, sampling initialization, and KPI dictionary setup.

```python
# ============================================================
# EXPERIMENT SETUP
# ============================================================
class Experiment:
    def __init__(self,
                 auditor,
                 random_number_set=DEFAULT_RND_SET,
                 n_streams=N_STREAMS,
                 iat=IAT,
                 # Keep misspelled arg for backward compatibility with existing cells.
                 traige_rejected=PCT_REFERRAL_REJECTED,
                 triage_rejected=None,
                 screening_discharge=PCT_SCREENING_DISCHARGED,
                 pre_assessment_rejection=PCT_PRE_ASS_REJECTED,
                 # Assessment outcome splits
                 pct_non_diag_at_assessment=PCT_NON_DIAGNOSIS_AT_ASSESSMENT,
                 pct_non_diag_at_further_assessment=PCT_NON_DIAGNOSIS_AT_FURTHER_ASSESSMENT,
                 # Post-diagnostic pathway
                 pct_post_diag_clinical=PCT_POST_DIAG_CLINICAL,
                 pct_removal_discharge=PCT_REMOVAL_DISCHARGE,
                 # Durations
                 dur_screening=DURATION_SCREENING,
                 dur_pre_assessment=DURATION_PRE_ASSESSMENT,
                 dur_assessment=DURATION_ASSESSMENT,
                 dur_further_assessment=DURATION_FURTHER_ASSESSMENT,
                 dur_post_diag_clinical=DURATION_POST_DIAG_CLINICAL,
                 dur_post_diag_other=DURATION_POST_DIAG_OTHER,
                 dur_review=DURATION_REVIEW,
                 # Staff / capacity
                 staff_screening=STAFF_SCREENING,
                 staff_pre_assessment=STAFF_PRE_ASSESSMENT,
                 staff_assessment=STAFF_ASSESSMENT,
                 staff_further_assessment=STAFF_FURTHER_ASSESSMENT,
                 staff_post_diag_clinical=STAFF_POST_DIAG_CLINICAL,
                 staff_post_diag_other=STAFF_POST_DIAG_OTHER,
                 staff_review=STAFF_REVIEW,
                 ):
        self.auditor = auditor
        self.random_number_set = random_number_set
        self.n_streams = n_streams

        self.iat = iat
        self.traige_rejected = traige_rejected if triage_rejected is None else triage_rejected
        self.screening_discharge = screening_discharge
        self.pre_assessment_rejection = pre_assessment_rejection

        self.pct_non_diag_at_assessment = pct_non_diag_at_assessment
        self.pct_non_diag_at_further_assessment = pct_non_diag_at_further_assessment

        self.pct_post_diag_clinical = pct_post_diag_clinical
        self.pct_removal_discharge = pct_removal_discharge

        self.dur_screening = dur_screening
        self.dur_pre_assessment = dur_pre_assessment
        self.dur_assessment = dur_assessment
        self.dur_further_assessment = dur_further_assessment
        self.dur_post_diag_clinical = dur_post_diag_clinical
        self.dur_post_diag_other = dur_post_diag_other
        self.dur_review = dur_review

        self.staff_screening = staff_screening
        self.staff_pre_assessment = staff_pre_assessment
        self.staff_assessment = staff_assessment
        self.staff_further_assessment = staff_further_assessment
        self.staff_post_diag_clinical = staff_post_diag_clinical
        self.staff_post_diag_other = staff_post_diag_other
        self.staff_review = staff_review

        self.init_results_variables()
        self.init_sampling()

    def set_random_no_set(self, random_number_set):
        self.random_number_set = random_number_set
        self.init_sampling()

    def init_sampling(self):
        seed_sequence = np.random.SeedSequence(self.random_number_set)
        self.seeds = seed_sequence.spawn(self.n_streams)

        def to_days(hours_list):
            return [h / 24.0 for h in hours_list]

        self.iat_dist = Exponential(self.iat, random_seed=self.seeds[0])
        self.screening_time_dist = Triangular(*to_days(self.dur_screening), random_seed=self.seeds[1])
        self.pre_ass_time_dist = Triangular(*to_days(self.dur_pre_assessment), random_seed=self.seeds[2])
        self.assessment_time_dist = Triangular(*to_days(self.dur_assessment), random_seed=self.seeds[3])
        self.referral_reject_dist = Bernoulli(self.traige_rejected, random_seed=self.seeds[4])
        self.screening_discharge_dist = Bernoulli(self.screening_discharge, random_seed=self.seeds[5])
        self.pre_ass_reject_dist = Bernoulli(self.pre_assessment_rejection, random_seed=self.seeds[6])
        self.non_diag_at_assessment_dist = Bernoulli(self.pct_non_diag_at_assessment, random_seed=self.seeds[7])
        self.further_assessment_time_dist = Triangular(*to_days(self.dur_further_assessment), random_seed=self.seeds[8])
        self.non_diag_at_further_assessment_dist = Bernoulli(
            self.pct_non_diag_at_further_assessment, random_seed=self.seeds[9]
        )

        self.post_diag_clinical_dist = Bernoulli(self.pct_post_diag_clinical, random_seed=self.seeds[10])
        self.post_diag_clinical_time_dist = Triangular(*to_days(self.dur_post_diag_clinical), random_seed=self.seeds[11])
        self.post_diag_other_time_dist = Triangular(*to_days(self.dur_post_diag_other), random_seed=self.seeds[12])
        self.review_time_dist = Triangular(*to_days(self.dur_review), random_seed=self.seeds[13])
        self.removal_discharge_dist = Bernoulli(self.pct_removal_discharge, random_seed=self.seeds[14])

    def init_results_variables(self):
        self.results = {
            # System level
            "ARRIVED_TOTAL": 0,
            "EXIT_TOTAL": 0,
            "IN_SYSTEM_END": 0,
            "CLINICAL_COMPLETED_TOTAL": 0,
            "SYSTEM_COMPLETED_TOTAL": 0,

            # Referral / triage
            "ARRIVED_REFERRAL": 0,
            "FLOW_REFERRAL_ACCEPTED": 0,
            "EXIT_REFERRAL_REJECTED": 0,

            # Screening
            "ARRIVED_SCREENING": 0,
            "SERVICE_SCREENING_COMPLETED": 0,
            "FLOW_SCREENING_PASSED": 0,
            "EXIT_SCREENING_DISCHARGED": 0,

            # Pre-assessment
            "ARRIVED_PRE_ASSESS": 0,
            "SERVICE_PRE_ASSESS_COMPLETED": 0,
            "FLOW_PRE_ASSESS_PASSED": 0,
            "EXIT_PRE_ASSESS_REJECTED": 0,

            # Assessment
            "ARRIVED_ASSESSMENT": 0,
            "SERVICE_ASSESSMENT_COMPLETED": 0,
            "FLOW_ASSESSMENT_PASSED": 0,
            "EXIT_ASSESSMENT_NON_DIAGNOSIS": 0,

            # Further assessment
            "ARRIVED_FURTHER_ASSESS": 0,
            "SERVICE_FURTHER_ASSESS_COMPLETED": 0,
            "FLOW_FURTHER_ASSESS_PASSED": 0,
            "EXIT_FURTHER_NON_DIAGNOSIS": 0,

            # Diagnosis
            "FLOW_DIAGNOSIS_CONFIRMED": 0,

            # Post-diagnostic support
            "FLOW_POST_DIAG_CLINICAL_ACCEPTED": 0,
            "SERVICE_POST_DIAG_CLINICAL_COMPLETED": 0,
            "FLOW_POST_DIAG_OTHER_ACCEPTED": 0,
            "SERVICE_POST_DIAG_OTHER_COMPLETED": 0,

            # Discharge / exit
            "EXIT_FORMAL_DISCHARGE": 0,
            "EXIT_SELF_REMOVED": 0,
        }
```

## 7. Patient Pathway Logic

Stage-by-stage patient flow, queueing, branching, and exits.

```python
class Patient:
    def __init__(self, patient_id, system):
        self.patient_id = patient_id
        self.system = system
        self.env = system.env
        self.args = system.args

    def process(self):
        self.args["ARRIVED_TOTAL"] += 1
        self.args["ARRIVED_REFERRAL"] += 1

        # Referral triage
        if self.system.referral_reject_dist.sample() == 1:
            self.args["EXIT_REFERRAL_REJECTED"] += 1
            self.args["EXIT_TOTAL"] += 1
            return
        self.args["FLOW_REFERRAL_ACCEPTED"] += 1

        # Screening
        self.args["ARRIVED_SCREENING"] += 1
        with self.system.screening_resource.request() as req:
            yield req
            yield self.env.timeout(self.system.screening_time_dist.sample())
            self.args["SERVICE_SCREENING_COMPLETED"] += 1

        if self.system.screening_discharge_dist.sample() == 1:
            self.args["EXIT_SCREENING_DISCHARGED"] += 1
            self.args["EXIT_TOTAL"] += 1
            return
        self.args["FLOW_SCREENING_PASSED"] += 1

        # Pre-assessment
        self.args["ARRIVED_PRE_ASSESS"] += 1
        with self.system.pre_assessment_resource.request() as req:
            yield req
            yield self.env.timeout(self.system.pre_ass_time_dist.sample())
            self.args["SERVICE_PRE_ASSESS_COMPLETED"] += 1

        if self.system.pre_ass_reject_dist.sample() == 1:
            self.args["EXIT_PRE_ASSESS_REJECTED"] += 1
            self.args["EXIT_TOTAL"] += 1
            return
        self.args["FLOW_PRE_ASSESS_PASSED"] += 1

        # Assessment
        self.args["ARRIVED_ASSESSMENT"] += 1
        with self.system.assessment_resource.request() as req:
            yield req
            yield self.env.timeout(self.system.assessment_time_dist.sample())
            self.args["SERVICE_ASSESSMENT_COMPLETED"] += 1

        if self.system.non_diag_at_assessment_dist.sample() == 1:
            self.args["EXIT_ASSESSMENT_NON_DIAGNOSIS"] += 1
            self.args["EXIT_TOTAL"] += 1
            return
        self.args["FLOW_ASSESSMENT_PASSED"] += 1

        # Further assessment
        self.args["ARRIVED_FURTHER_ASSESS"] += 1
        with self.system.further_assessment_resource.request() as req:
            yield req
            yield self.env.timeout(self.system.further_assessment_time_dist.sample())
            self.args["SERVICE_FURTHER_ASSESS_COMPLETED"] += 1

        if self.system.non_diag_at_further_assessment_dist.sample() == 1:
            self.args["EXIT_FURTHER_NON_DIAGNOSIS"] += 1
            self.args["EXIT_TOTAL"] += 1
            return

        self.args["FLOW_FURTHER_ASSESS_PASSED"] += 1
        self.args["FLOW_DIAGNOSIS_CONFIRMED"] += 1

        # Post-diagnostic support
        if self.system.post_diag_clinical_dist.sample() == 1:
            self.args["FLOW_POST_DIAG_CLINICAL_ACCEPTED"] += 1
            with self.system.post_diag_clinical_resource.request() as req:
                yield req
                yield self.env.timeout(self.system.post_diag_clinical_time_dist.sample())
                self.args["SERVICE_POST_DIAG_CLINICAL_COMPLETED"] += 1
        else:
            self.args["FLOW_POST_DIAG_OTHER_ACCEPTED"] += 1
            with self.system.post_diag_other_resource.request() as req:
                yield req
                yield self.env.timeout(self.system.post_diag_other_time_dist.sample())
                self.args["SERVICE_POST_DIAG_OTHER_COMPLETED"] += 1

        # Discharge / removal
        with self.system.review_resource.request() as req:
            yield req
            yield self.env.timeout(self.system.review_time_dist.sample())

        if self.system.removal_discharge_dist.sample() == 1:
            self.args["EXIT_FORMAL_DISCHARGE"] += 1
        else:
            self.args["EXIT_SELF_REMOVED"] += 1

        self.args["CLINICAL_COMPLETED_TOTAL"] += 1
        self.args["EXIT_TOTAL"] += 1
```

## 8. System Resource Setup

System class wiring distributions, resources, and patient generation.

```python
class AutismPathwaySystem:
    def __init__(self, env, experiment, arrival_stop=None):
        self.env = env
        self.experiment = experiment
        self.args = experiment.results
        self.active_patients_list = []
        self.arrival_stop = arrival_stop

        # Distributions and RNG streams from experiment
        self.iat_dist = experiment.iat_dist
        self.screening_time_dist = experiment.screening_time_dist
        self.pre_ass_time_dist = experiment.pre_ass_time_dist
        self.assessment_time_dist = experiment.assessment_time_dist
        self.referral_reject_dist = experiment.referral_reject_dist
        self.screening_discharge_dist = experiment.screening_discharge_dist
        self.pre_ass_reject_dist = experiment.pre_ass_reject_dist
        self.non_diag_at_assessment_dist = experiment.non_diag_at_assessment_dist
        self.further_assessment_time_dist = experiment.further_assessment_time_dist
        self.non_diag_at_further_assessment_dist = experiment.non_diag_at_further_assessment_dist
        self.post_diag_clinical_dist = experiment.post_diag_clinical_dist
        self.post_diag_clinical_time_dist = experiment.post_diag_clinical_time_dist
        self.post_diag_other_time_dist = experiment.post_diag_other_time_dist
        self.review_time_dist = experiment.review_time_dist
        self.removal_discharge_dist = experiment.removal_discharge_dist

        # Staff resources sourced from experiment parameters
        self.screening_resource = simpy.Resource(env, capacity=experiment.staff_screening)
        self.pre_assessment_resource = simpy.Resource(env, capacity=experiment.staff_pre_assessment)
        self.assessment_resource = simpy.Resource(env, capacity=experiment.staff_assessment)
        self.further_assessment_resource = simpy.Resource(env, capacity=experiment.staff_further_assessment)
        self.post_diag_clinical_resource = simpy.Resource(env, capacity=experiment.staff_post_diag_clinical)
        self.post_diag_other_resource = simpy.Resource(env, capacity=experiment.staff_post_diag_other)
        self.review_resource = simpy.Resource(env, capacity=experiment.staff_review)

        # Start arrival generator process
        self.env.process(self.generate_patients())

    def generate_patients(self):
        for patient_id in itertools.count(1):
            inter_arrival_time = self.iat_dist.sample()
            next_arrival = self.env.now + inter_arrival_time

            # Stop creating new arrivals after arrival_stop horizon.
            if self.arrival_stop is not None and next_arrival > self.arrival_stop:
                break

            yield self.env.timeout(inter_arrival_time)
            new_patient = Patient(patient_id, self)
            self.env.process(new_patient.process())
            self.active_patients_list.append(new_patient)
            trace(f"Patient {patient_id} arrived at time {self.env.now:.2f} days.")
```

## 9. Single Run Function

Run one replication and return results.

```python
def single_run(experiment, rep=0, run_length=RUN_LENGTH):
    """
    Perform a single run of the model and return the results.

    Arrivals are generated only up to run_length, then the model drains
    until all arrived patients complete/exit.
    """
    experiment.init_results_variables()
    experiment.set_random_no_set(rep)

    env = simpy.Environment()
    AutismPathwaySystem(env, experiment, arrival_stop=run_length)

    # Drain model so conservation equations are exact at run end.
    env.run()
    trace(f"Rep {rep}: Simulation run completed.")

    experiment.results["IN_SYSTEM_END"] = (
        experiment.results["ARRIVED_TOTAL"] - experiment.results["EXIT_TOTAL"]
    )
    experiment.results["SYSTEM_COMPLETED_TOTAL"] = experiment.results["EXIT_TOTAL"]

    results = experiment.results.copy()
    return results
```

## 13. Stage Boundary Tests (100% exits and 0% exits)

This section checks that pathway branching works as intended at each stage.

### Why this matters

The model uses stage probabilities to decide whether a patient exits or moves forward. These tests confirm that the logic is correct at the two extreme settings.

### What we test

1. Stages tested: triage, screening, pre-assessment, assessment, further assessment.
2. Boundary settings tested at each stage:
   - `1.0` (100% exit at that stage)
   - `0.0` (0% exit at that stage)

### How we test

1. Run the reusable boundary test function for each stage and boundary setting.
2. For `1.0` (100% exit), check:
   - exits at stage = arrivals at stage
   - pass-through = 0
   - downstream entry = 0
3. For `0.0` (0% exit), check:
   - exits at stage = 0
   - arrivals at stage > 0
   - pass-through = arrivals at stage
   - downstream entry > 0

### Link to model logic

These checks map directly to stage transitions in `Patient.process()` and KPI counters in `Experiment.results`. If all checks pass, the stage-gate behavior is consistent with the model design.

## 14. Replications and Summary Statistics

Run multiple replications and summarize central tendency and uncertainty.

```python
def multiple_runs(experiment, n_reps=N_REP, 
                  run_length=RUN_LENGTH, 
                  n_jobs=-1):
    """
    Run multiple simulation replications in parallel using joblib.

    Parameters
    ----------
    experiment : Experiment
        The experiment/parameters to use with model.
    n_reps : int
        Number of simulation replications.
    run_length : float
        Length of each simulation run in days.
    n_jobs : int
        Number of CPU cores to use.
        -1 = use all available cores.

    Returns
    -------
    pd.DataFrame
        DataFrame containing results from all replications.
    """

    # Run simulations in parallel
    all_results = Parallel(n_jobs=n_jobs)(
        delayed(single_run)(
            experiment=experiment,
            rep=rep,
            run_length=run_length
        )
        for rep in range(n_reps)
    )

    # Add replication number
    for rep, result in enumerate(all_results):
        result["Replication"] = rep

    # Convert results to dataframe
    results_df = pd.DataFrame(all_results)

    return results_df
```

```python
# =========================================================
# RUN MULTIPLE SIMULATION REPLICATIONS
# =========================================================

# Example: run 8 replications with default parameters
experiment = Experiment(auditor=Audit())
n_reps = 6
results_df = multiple_runs(experiment, 
                           n_reps=n_reps, 
                           run_length=RUN_LENGTH,
                           n_jobs=-1)
results_df
```

```python

def summarise_runs(results_df):
    """
    Create summary statistics for simulation replications.
    """

    numeric_cols = results_df.select_dtypes(include=np.number).columns

    summary = pd.DataFrame({
        "Mean": results_df[numeric_cols].mean(),
        "Std Dev": results_df[numeric_cols].std(),
        "Min": results_df[numeric_cols].min(),
        "Max": results_df[numeric_cols].max()
    })

    # 95% confidence interval
    n = len(results_df)

    summary["95% CI Lower"] = (
        summary["Mean"] -
        1.96 * summary["Std Dev"] / np.sqrt(n)
    )

    summary["95% CI Upper"] = (
        summary["Mean"] +
        1.96 * summary["Std Dev"] / np.sqrt(n)
    )

    return summary.round(2)
summary_stats = summarise_runs(results_df)
summary_stats
```
